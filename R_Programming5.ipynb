```r
# Missing value
x <- NA

print(x)

# Check whether x is missing
is.na(x)
```

```r
x <- c(11, NA, 13, NA)

print(x)

# Check missing values
is.na(x)
```

```r
x <- c(11, NA, 13, NA)

# Mean without removing NA
mean(x)

# Mean after removing NA
mean(x, na.rm = TRUE)
```
